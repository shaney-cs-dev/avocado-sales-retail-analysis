```python
import pandas as pd

# 1. Load the Excel file using the exact name found
df = pd.read_excel('avocado.csv.xlsx')

# 2. Drop the redundant index column if it exists
if 'Unnamed: 0' in df.columns:
    df = df.drop(columns=['Unnamed: 0'])

# 3. Rename generic columns to make them human-readable
df = df.rename(columns={
    'Total Volume': 'total_volume',
    '4046': 'small_hass_sold',
    '4225': 'medium_hass_sold',
    '4770': 'large_hass_sold',
    'Total Bags': 'total_bags',
    'Small Bags': 'small_bags',
    'Large Bags': 'large_bags',
    'XLarge Bags': 'xlarge_bags',
    'AveragePrice': 'average_price'
})

# 4. Standardize text values to lowercase
df['type'] = df['type'].str.lower()
df['region'] = df['region'].str.lower()

# 5. Correct the date data type and pull out month/year features
df['Date'] = pd.to_datetime(df['Date'])
df['month'] = df['Date'].dt.month
df['year'] = df['Date'].dt.year

# 6. Verify data integrity checks
print(f"Missing Values Found:\n{df.isnull().sum()}\n")
print(f"Total Duplicate Rows: {df.duplicated().sum()}\n")
print("--- Cleaned Data Preview ---")
print(df[['Date', 'average_price', 'total_volume', 'type', 'region']].head(3))

# 7. Save this pristine version to a new file
df.to_csv('avocado_cleaned.csv', index=False)
print("\nSuccess! 'avocado_cleaned.csv' has been generated in your file panel.")
```

```python
import pandas as pd

# Load our freshly cleaned data
df = pd.read_csv('avocado_cleaned.csv')

print("=== 1. CONVENTIONAL VS ORGANIC SUMMARY ===")
# Compare overall average prices and sales volume between types
type_summary = df.groupby('type')[['average_price', 'total_volume']].agg({'average_price': 'mean', 'total_volume': 'sum'})
print(type_summary)

print("\n=== 2. TOP 5 MOST EXPENSIVE REGIONS ===")
# Find which regions have the highest average avocado prices
expensive_regions = df.groupby('region')['average_price'].mean().sort_values(ascending=False).head(5)
print(expensive_regions)

print("\n=== 3. SEASONAL TRENDS (BY MONTH) ===")
# Check monthly price fluctuations to find peak seasonal prices
monthly_trends = df.groupby('month')['average_price'].mean()
print(monthly_trends)

print("\n=== 4. PRICE ELASTICITY (CORRELATION) ===")
# Measure the relationship between price and total volume sold
correlation = df['average_price'].corr(df['total_volume'])
print(f"Correlation coefficient between price and volume: {correlation:.4f}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the clean dataset
df = pd.read_csv('avocado_cleaned.csv')

# Format the visualization style
plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")

# Create a line chart showing seasonal price patterns across the months
sns.lineplot(x='month', y='average_price', data=df, marker='o', color='forestgreen', linewidth=2.5)

# Add titles and labels for stakeholder readability
plt.title('US Avocado Average Price Fluctuations by Month', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Month (1 = January, 12 = December)', fontsize=12)
plt.ylabel('Average Price ($)', fontsize=12)
plt.xticks(range(1, 13))

# Display the final chart
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the clean dataset
df = pd.read_csv('avocado_cleaned.csv')

# Aggregate total volume by type
volume_by_type = df.groupby('type')['total_volume'].sum().reset_index()

# Format the visualization style
plt.figure(figsize=(8, 6))
sns.set_theme(style="whitegrid")

# Create a side-by-side bar plot
ax = sns.barplot(
    x='type',
    y='total_volume',
    data=volume_by_type,
    palette=['#2b5c8f', '#4e8a44']  # Professional blue and organic green tints
)

# Format the y-axis to show billions clearly instead of scientific notation
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f'{x*1e-9:.1f}B'))

# Add titles and labels for stakeholder readability
plt.title('Total Sales Volume Comparison: Conventional vs. Organic', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Avocado Type', fontsize=12)
plt.ylabel('Total Volume Sold (in Billions)', fontsize=12)

# Add data labels on top of the bars
for p in ax.patches:
    ax.annotate(f'{p.get_height()*1e-9:.2f}B',
                (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='center',
                xytext=(0, 9),
                textcoords='offset points',
                fontsize=11, fontweight='bold')

# Display the final chart
plt.tight_layout()
plt.show()
```
